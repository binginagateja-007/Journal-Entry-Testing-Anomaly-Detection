```python
import pandas as pd

scored = pd.read_csv("journal_entries_scored.csv")
vendor = pd.read_csv("vendor_risk_summary.csv")
dept = pd.read_csv("department_risk_summary.csv")

with pd.ExcelWriter("JE_Testing_PowerBI_Data.xlsx", engine="openpyxl") as writer:
    scored.to_excel(writer, sheet_name="Transactions_Scored", index=False)
    vendor.to_excel(writer, sheet_name="Vendor_Risk_Summary", index=False)
    dept.to_excel(writer, sheet_name="Department_Risk_Summary", index=False)

print("done", scored.shape, vendor.shape, dept.shape)
```
